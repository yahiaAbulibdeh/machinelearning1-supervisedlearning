# **2. Import the necessary pakeges, classes, and functions**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.linear_model import SGDRegressor
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
```

# **4. Read and explore the dataset**

```python
insurance_data=pd.read_csv("insurance.csv")
insurance_data.head() # returns the first 5 rows with their columns
```

```python
insurance_data.shape # return number of rows and columns
```

# **5. Split the dataset to Independent and Dependent *variables*:**
  Independent variables(x) and Dependent Variable (y)

```python
x=insurance_data.drop("charges", axis=1)
y=insurance_data["charges"]
```

```python
x.shape
```

```python
x.head()
```

# **6. Depict the correlation between Smoking and the average charges**
The type of plot depends on the type of column to be depicted

```python
plt.bar(x["smoker"], y, width=0.5)
plt.title('Correlation Between Smoking & Charges')
plt.xticks([0, 1], ["Not Smoker", "Smoker"])
plt.show()
```

# **7. Split the dataset to train and test data:**
Test set size=20% of the whole dataset

```python
x_train, x_test, y_train, y_test=train_test_split(x, y, test_size=0.20)
```

```python
x_train.shape
```

```python
x_test.shape
```

# **8. Create Linear Regression model based on OLS**

## 8.1 Construct the model

```python
ols=LinearRegression()
```

## 8.2 Train the model

```python
ols.fit(x_train,y_train)
```

## 8.3 Model Prediction

```python
ols_pred=ols.predict(x_test)
```

## 8.4 Model Evaluation

```python
print(f" R^2= {r2_score(y_test,ols_pred):0.2f}")
print(f"MAE= {mean_absolute_error(y_test,ols_pred):0.2f}")
print(f"MSE= {mean_squared_error(y_test,ols_pred):0.2f}")
```

# **9.  Create Linear Regression model based on GD**

## 9.1 Preprocessing
using StandardScaler to scale the independent variables value

```python
scaler=StandardScaler()
new_x_train=scaler.fit_transform(x_train)
new_x_test=scaler.fit_transform(x_test)
```

```python
new_x_train # to display tha values of new_x_train, it transformed to 2d array
```

## 9.2 Construct the model

```python
gd=SGDRegressor()
```

## 9.3 Train the model

```python
gd.fit(new_x_train,y_train)
```

## 9.4 Model Prediction

```python
gd_pred=gd.predict(new_x_test)
```

## 9.5  Model Evaluation

```python
print(f" R^2= {r2_score(y_test,gd_pred):0.2f}")
print(f"MAE= {mean_absolute_error(y_test,gd_pred):0.2f}")
print(f"MSE= {mean_squared_error(y_test,gd_pred):0.2f}")
```
