```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix , classification_report , accuracy_score
```

```python
my_data = pd.read_csv('heart_failure.csv')
```

```python
my_data.head()
```

```python
my_data.shape
```

```python
x = my_data.drop('DEATH_EVENT' , axis = 1)
y = my_data['DEATH_EVENT']
```

```python
 x_train , x_test , y_train , y_test = train_test_split(x , y , test_size = 0.2 , random_state = 12)
```

```python
  x_test.shape
```

```python
log_model = LogisticRegression()
```

```python
log_model.fit(x_train ,y_train)
```

```python
log_pred = log_model.predict(x_test)
```

```python
conf_matrix = confusion_matrix(y_test , log_pred)
```

```python
conf_matrix
```

```python
plt.title('Confusion Matrix' , pad = '20')
plt.xticks([0,1])
plt.yticks([0,1])
sns.heatmap(conf_matrix , annot = True , cmap = 'YlGnBu')
plt.xlabel('Predicion Lables')
plt.ylabel('Actioal laples')
```

```python
c = ['survived' , 'died']
report = classification_report(y_test , log_pred , target_names = c)
print (report)
```

```python
print(accuracy_score(y_test , log_pred))
```
