# **1. Independent and Dependent Varaibles**

```python
import numpy as np
x=[[1,2,1,1],[1,4,2,2],[1,4,3,1],[1,3,1,3],[1,1,1,1]] # 2d array of training data (Independent Varaibles)
y=[[13],[20],[22],[14],[11]] # 2d array of  Dependent Variable
```

# **2. Cost Function**

```python
def cost_Func(x,y,w):
  y_hat=np.matmul(x,w)# find the predicted values
  error=y_hat-y # find the difference between the predicted and actual values
  sqr_error=np.square(error)
  mse=np.mean(sqr_error)
  return mse
```

# **3. Coefficients of Partial Derivatives and Constants**

```python
delta_c=[[1,2.8,1.6,1.6],[2.8,9.2,5.2,4.8],[1.6,5.2,3.2,2.4],[1.6,4.8,2.4,3.2]]# 2d array of Partial Derivatives Coefficients
delta_c_y=[[-16],[-49],[-29],[-26]]# 2d array of Partial Derivatives  Constants
```

# **4. Gradient Descent Function and Optimize MSE**

```python
def gd(x,y,delta_c,delta_c_y):
  w=[[4],[1],[2],[0]]
  Mean_Sqr_error=cost_Func(x,y,w)
  print("Initial values of w: ",w)
  print("Initial value of MSE: ",Mean_Sqr_error)
  print("---------------")
  for i in range(0,5): # to   optimize MSE
    partial_derivative_values=np.matmul(delta_c,w)+delta_c_y # find the partial derivative values with respect to w values
    w=w-0.05*partial_derivative_values # optimize w values, 0.05 is the learning rate
    Mean_Sqr_error=cost_Func(x,y,w)
    print("w iteration# ",i,np.round(w,4))
    print("MSE: ",np.round(Mean_Sqr_error,4))
    print("---------------")
```

```python
gd(x,y,delta_c,delta_c_y)
```
