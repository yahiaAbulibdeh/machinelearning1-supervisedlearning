```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn import tree
from sklearn.model_selection import train_test_split, cross_validate, GridSearchCV
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import OrdinalEncoder , OneHotEncoder
from sklearn.metrics import confusion_matrix, classification_report, accuracy_score
```

```python
car_evaluation_data = pd.read_csv("car_evaluation.csv", header=None)
```

```python
car_evaluation_data.shape
```

```python
car_evaluation_data.head()
```

```python
column_names = ['Buying Price', 'Maintenance Cost', 'Doors', 'Persons', ' Luggage Boot Size', 'Safety', 'Class']
```

```python
car_evaluation_data.columns = column_names
```

```python
car_evaluation_data.head(5)
```

```python
car_evaluation_data.isna().sum()
```

```python
X = car_evaluation_data.drop("Class", axis = 1)
y = car_evaluation_data["Class"]
```

```python
X.head()
```

```python
X.shape
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.25 , random_state = 42)
```

```python
X_test.shape
```

```python
encoder = OrdinalEncoder(categories=[['low', 'med', 'high', 'vhigh'],
                                     ['low', 'med', 'high', 'vhigh'],
                                     ['2', '3', '4', '5more'],
                                     ['2', '4', 'more'],
                                     ['small', 'med', 'big'],
                                     ['low', 'med', 'high']])
```

```python
X_train = encoder.fit_transform(X_train)
X_test = encoder.fit_transform(X_test)
```

```python
X_train = pd.DataFrame(X_train)
X_test = pd.DataFrame(X_test)
```

```python
X_train.head()
```

```python
column_names.remove('Class')
```

```python
X_train.columns = column_names
X_test.columns = column_names
```

```python
X_train.head()
```

```python
X_test.head()
```

```python
logistic_model = LogisticRegression(max_iter=100)
```

```python
logistic_model_results = cross_validate(logistic_model, X_train, y_train, cv= 5)
```

```python
print('Results:', logistic_model_results)
```

```python
print('Test Scores:', logistic_model_results['test_score'])
```

```python
print('Average Score:', format(logistic_model_results['test_score'].mean(), '.1%'))
```

```python
logistic_model = LogisticRegression(max_iter=100)
```

```python
logistic_model.fit(X_train, y_train)
```

```python
logistic_model_preds = logistic_model.predict(X_test)
```

```python
print(format(accuracy_score(y_test, logistic_model_preds), ".1%"))
```

```python
knn_model = KNeighborsClassifier(n_neighbors=5)
```

```python
knn_model_results = cross_validate(knn_model, X_train, y_train, cv=5)
```

```python
print('Results:', knn_model_results)
```

```python
print('Test Scores:', knn_model_results['test_score'])
```

```python
print('Average Scores:', format(knn_model_results['test_score'].mean(), '.1%'))
```

```python
knn_model = KNeighborsClassifier(n_neighbors=5)
```

```python
knn_model.fit(X_train, y_train)
```

```python
knn_model_preds = knn_model.predict(X_test)
```

```python
print(format(accuracy_score(y_test, knn_model_preds), ".1%"))
```

```python
decision_tree_parameters = {'criterion': ['gini', 'entropy'], 'max_depth': [3, 4, 5, 6, 7, 8, 9, 10]}
```

```python
decision_tree_grid = GridSearchCV(DecisionTreeClassifier(), decision_tree_parameters, cv=5)
```

```python
decision_tree_grid.fit(X_train, y_train)
```

```python
print(decision_tree_grid.best_params_)
```

```python
print(format(decision_tree_grid.best_score_, '.1%'))
```

```python
decision_tree_model = DecisionTreeClassifier(criterion='entropy', max_depth=10, random_state=42)
```

```python
decision_tree_model.fit(X_train, y_train)
```

```python
decision_tree_model_preds = decision_tree_model.predict(X_test)
```

```python
print(format(accuracy_score(y_test, decision_tree_model_preds), ".1%"))
```

```python
random_forest_parameters = {'criterion': ['gini', 'entropy'], 'n_estimators': [25, 50, 100, 150, 250, 500], 'max_depth': [3, 4, 5, 6, 7, 8, 9, 10]}
```

```python
random_forest_grid = GridSearchCV(RandomForestClassifier(random_state=42), random_forest_parameters, cv=5)
```
